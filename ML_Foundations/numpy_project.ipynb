```python
import numpy as np
```

```python
data = np.genfromtxt(
    "real_estate_dataset.csv",
    delimiter=",",
    names=True,
    dtype=None,
    encoding="utf-8"
)
```

```python
price = data["Price"]

print("Mean:", np.mean(price))
print("Median:", np.median(price))
print("Minimum:", np.min(price))
print("Maximum:", np.max(price))
```

```python
price_range = np.max(price) - np.min(price)

print("Price range:", price_range)
```

```python
print("25th percentile:", np.percentile(price, 25))
print("50th percentile:", np.percentile(price, 50))
print("75th percentile:", np.percentile(price, 75))
```

```python
square_feet = data["Square_Feet"]
```

```python
print("Mean:", np.mean(square_feet))
print("Median:", np.median(square_feet))
print("Minimum:", np.min(square_feet))
print("Maximum:", np.max(square_feet))
```

```python
price_per_sqft = price / square_feet
```

```python
print(price_per_sqft[:10])
```

```python
print("Mean:", np.mean(price_per_sqft))
print("Median:", np.median(price_per_sqft))
print("Minimum:", np.min(price_per_sqft))
print("Maximum:", np.max(price_per_sqft))
```

```python
expensive_mask = price > np.percentile(price, 75)
```

```python
print(expensive_mask[:10])
```

```python
expensive_prices = price[expensive_mask]

print(expensive_prices)
```

```python
print("Number of expensive properties:", np.sum(expensive_mask))
```

```python
print("Average price of expensive properties:",
      np.mean(expensive_prices))
```

```python
print("Overall average price:",
      np.mean(price))
```

```python
large_mask = square_feet > np.mean(square_feet)
```

```python
location_score = data["Location_Score"]

well_located_mask = location_score >= 8
```

```python
large_expensive_well_located = (
    large_mask &
    expensive_mask &
    well_located_mask
)
```

```python
print(np.sum(large_expensive_well_located))
```

```python
selected = data[large_expensive_well_located]

print(selected)
```

```python
print("IDs:", selected["ID"])
print("Size:", selected["Square_Feet"])
print("Location Score:", selected["Location_Score"])
print("Price:", selected["Price"])
```
